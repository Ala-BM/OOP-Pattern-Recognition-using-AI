```python
def textPro(txt,csv,pkl):
        import re
        with open(txt, "r") as file:
            content = file.read()

        CoPregex = r"(.*) is a Chain of Responsibility Handler class\n.* is a handle operation\n.* of type (.*) propogates the request\nFile Location:.*"
        CoP = re.findall(CoPregex, content)
        CoP = list(dict.fromkeys(CoP))
        MediatorRegex =r"Mediator Pattern.\nMediator: (.*)\nColleagues: (.*)\nFileLocation:.*"
        Mediator = re.findall(MediatorRegex, content)
        for i,k in enumerate(Mediator):
           Mediator[i]=(Mediator[i][0],*Mediator[i][1].split())
        Mediator = list(dict.fromkeys(Mediator))

        ObserverRegex =r"Observer Pattern.\n(.*) is an observer iterator.\n(.*) is the generic type for the listeners.\n.* is the notify method.\n.* is the update method.\nSubject classes: (.*)\nFile Location:.*"
        Observer = re.findall(ObserverRegex,content)
        for i,k in enumerate(Observer):
           Observer[i]=(Observer[i][0],Observer[i][1],*Observer[i][2].split())
        Observer = list(dict.fromkeys(Observer))

        StateRegex = r"State Pattern.\n(.*) is the Context class.\n(.*) is the State interface.\nConcrete State classes:  (.*)\nDelegation through .* of type .*\n.* changes the state variable .*\n.* is invoked by.*"
        State = re.findall(StateRegex,content)
        for i,k in enumerate(State):
           State[i]=(State[i][0],State[i][1],*State[i][2].split())
        State = list(dict.fromkeys(State))

        StrategyRegex=r"Strategy Pattern.\n(.*) is the Context class.\n(.*) is the Strategy interface.\nConcrete Strategy classes:  (.*)\n.*"
        Strategy = re.findall(StrategyRegex,content)
        for i,k in enumerate(Strategy):
           Strategy[i]=(Strategy[i][0],Strategy[i][1],*Strategy[i][2].split())
        Strategy = list(dict.fromkeys(Strategy))

        TMethodRegex=r"Template Method Found.\n(.*) is the template class\n.*is the template method\n.* is a primitive method.*"
        TMethod = re.findall(TMethodRegex,content)

        VisitorRegex=r"Visitor pattern found.\n(.*) is an abstract Visitor class.\n(.*) is a Vistee class.\n.* is the accept method.\n.* is the visit method.*"
        Visitor = re.findall(VisitorRegex,content)
        Visitor = list(dict.fromkeys(Visitor))

        AdapterRegex=r"Adapter Pattern.\nAdapting classes: (.*)\n(.*) is an adapter class.\n(.*) is the adaptee class.*"
        Adapter = re.findall(AdapterRegex,content)
        for i,k in enumerate(Adapter):
           Adapter[i]=(Adapter[i][1],Adapter[i][2],*Adapter[i][0].split())
        Adapter = list(dict.fromkeys(Adapter))

        BridgeRegex=r"Bridge Pattern.\n(.*) is abstract.\n(.*)is an interface.\n.*delegates .*\n.*"
        Bridge = re.findall(BridgeRegex,content)
        Bridge = list(dict.fromkeys(Bridge))

        CompositeRegex=r"Composite pattern.\n(.*)is the composite class.\n.* is the composite instance.\n(.*) is the component class."
        Composite = re.findall(CompositeRegex,content)
        Composite = list(dict.fromkeys(Composite))

        DecoratorRegex=r"Decorator Pattern\n(.*) is a Decorator class\n.* is a decorate operation\n.* of type (.*) is the Decoratee class.*"
        Decorator = re.findall(DecoratorRegex,content)
        Decorator = list(dict.fromkeys(Decorator))

        FacadeRegex=r"Facade Pattern.\n(.*) is a facade class.\nHidden types: .*\nFacade access types:.*\n.*"
        Facade = re.findall(FacadeRegex,content)
        Facade = list(dict.fromkeys(Facade))

        FlyweightRegex=r"Flyweight Pattern.\n(.*) is a flyweight factory.\n.* is a flyweight object.*"
        Flyweight = re.findall(FlyweightRegex,content)
        Flyweight = list(dict.fromkeys(Flyweight))

        ProxyRegex=r"Proxy Pattern.\n(.*) is a proxy.\n(.*) is a proxy interface.\nThe real objects: (.*)\n.*"
        Proxy = re.findall(ProxyRegex,content)
        Proxy = list(dict.fromkeys(Proxy))

        SingletonRegex=r"(.*) is a Singleton class\n.*"
        Singleton = re.findall(SingletonRegex,content)
        Singleton = list(dict.fromkeys(Singleton))

        FactoryMRegex=r"Factory Method pattern.\n(.*) is a Factory Method class.\n.* is a concrete Factory Method class.\n.* is a factory method returns  .*\n which extends (.*)\n.*"
        FactoryM = re.findall(FactoryMRegex,content)
        FactoryM = list(dict.fromkeys(FactoryM))

        import pandas as pd 
        Jdraw=pd.read_csv(csv)
        Jdraw.drop(['file','type','cbo','lcom','tcc','lcc','loc','returnQty','loopQty','comparisonsQty','tryCatchQty','parenthesizedExpsQty','stringLiteralsQty','numbersQty','assignmentsQty','mathOperationsQty','variablesQty','maxNestedBlocksQty','lambdasQty','uniqueWordsQty','modifiers','logStatementsQty'],inplace=True,axis=1)
        Jdraw.fillna(0)

        DBPatternX=[]
        DBPatternY=[]

        import statistics
        def meanLists(multiple_lists):
            mean=[statistics.mean(k) for k in zip(*multiple_lists)]
            print(mean,"IT IS WORKING")
            return mean

        def check_association (listX,listA):
         for name in listX:
            check = name[0].split("$")[0]
            for item in listA:
             if item==check:
                return name
         return listX[0]


        def extract_pattern (pattern_list,name,Jdraw,mod=0):
         match mod:
           case 0: 
            for item in pattern_list:
                Pattern=[]
                for classe in item: 
                  checks=(Jdraw.loc[(Jdraw['class'].str.contains('\\.'+classe+ '$')) | (Jdraw['class'].str.contains('\\$'+classe+ '$'))]).values.tolist()
                 ## if classe=='EventDispatcher':
                  ##  classe='AbstractCommand.EventDispatcher'         
                  if checks==[]:
                    print('RANDOM',classe,'\n')
                    Pattern=[]
                    break#we ingonring outside the source code classes TO BE FIXED
                  if len(checks)>1:
                        checks=[check_association(checks,item)]
                  Pattern.append(checks.pop(0)[1:])
                if Pattern==[]:
                 continue
                elif len(Pattern)==2:
                 listaM=sum(Pattern[0])+sum(Pattern[1])/len(Pattern[0])+len(Pattern[1])
                 Pattern.append([listaM]*29)
                DBPatternX.append(Pattern)
                DBPatternY.append(name)
           case 1:
             for item in pattern_list:
                 item=list(item) 
                 Pattern=[]   
                 means=[]
                 checks0=(Jdraw.loc[(Jdraw['class'].str.contains('\\.'+item[0]+ '$')) | (Jdraw['class'].str.contains('\\$'+item[0]+ '$'))]).values.tolist()
                 checks=(Jdraw.loc[(Jdraw['class'].str.contains('\\.'+item[1]+ '$')) | (Jdraw['class'].str.contains('\\$'+item[1]+ '$'))]).values.tolist()
                 if checks0==[]:
                    print('0',item[0],'\n')
                    continue
                 elif checks==[]:
                    print('REST',item[1],'\n')
                    continue
                 for i in range(2, len(item)-1):
                  checks2=(Jdraw.loc[(Jdraw['class'].str.contains('\\.'+item[i]+ '$')) | (Jdraw['class'].str.contains('\\$'+item[i]+ '$'))]).values.tolist()
                  if checks2==[]:
                    print('REST',item[i],'\n')
                    continue
                  elif len(checks2)>1:
                    checks2=[check_association(checks2,item)]
                  means.append(checks2.pop(0)[1:])
                 if (means==[]):
                    print("Mean empty")
                    continue
                 if len(checks)>1:
                        checks=[check_association(checks,item)]             
                 Pattern.append(checks0.pop(0)[1:])   
                 Pattern.append(checks.pop(0)[1:])
                 Pattern.append(meanLists(means))
                 DBPatternX.append(Pattern)
                 DBPatternY.append(name)

           case 2:
             for item in pattern_list:
                 item=list(item) 
                 Pattern=[]   
                 checks0=(Jdraw.loc[(Jdraw['class'].str.contains('\\.'+item[0]+ '$')) | (Jdraw['class'].str.contains('\\$'+item[0]+ '$'))]).values.tolist()
                 checks=(Jdraw.loc[(Jdraw['class'].str.contains('\\.'+item[1]+ '$')) | (Jdraw['class'].str.contains('\\$'+item[1]+ '$'))]).values.tolist()
                 checks2=(Jdraw.loc[(Jdraw['class'].str.contains('\\.'+item[2]+ '$')) | (Jdraw['class'].str.contains('\\$'+item[2]+ '$'))]).values.tolist()
                 if checks0==[]:
                    print('0',item[0],'\n')
                    continue
                 elif checks==[]:
                    print('REST',item[1],'\n')
                    continue
                 elif checks2==[]:
                    print('REST',item[2],'\n')
                    continue
                 if len(checks)>1:
                        checks=[check_association(checks,item)]   
                 elif len(checks2)>1:
                    checks2=[check_association(checks2,item)]
                 Pattern.append(checks0.pop(0)[1:])   
                 Pattern.append(checks.pop(0)[1:])
                 Pattern.append(checks2.pop(0)[1:])
                 DBPatternX.append(Pattern)
                 DBPatternY.append(name)
           case 3:
                for classe in pattern_list: 
                  Pattern=[]  
                  checks=(Jdraw.loc[(Jdraw['class'].str.contains('\\.'+classe+ '$')) | (Jdraw['class'].str.contains('\\$'+classe+ '$'))]).values.tolist()
                 ## if classe=='EventDispatcher':
                 ##   classe='AbstractCommand.EventDispatcher'
                  if len(checks)>1:
                        checks=[check_association(checks,pattern_list)]
                  if checks==[]:
                    print('RANDOM',classe,'\n')
                    continue
                  lista=checks.pop(0)[1:]
                  listaM=sum(lista)/len(lista)
                  Pattern.append(lista)
                  Pattern.append([listaM] * 29)
                  Pattern.append([listaM] * 29)
                  DBPatternX.append(Pattern)
                  DBPatternY.append(name)


        extract_pattern(CoP,"Chain_of_Responsibilities",Jdraw,0)
        #extract_pattern(Mediator,"Mediator",Jdraw,0)
        extract_pattern(Observer,"Observer",Jdraw,1)
        extract_pattern(State,"State",Jdraw,1)
        extract_pattern(Strategy,"Strategy",Jdraw,1)
        extract_pattern(TMethod,"Template_Method",Jdraw,3)
        extract_pattern(Visitor,"Visitor",Jdraw,0)
        extract_pattern(Adapter,"Adapter",Jdraw,1)
        extract_pattern(Bridge,"Bridge",Jdraw,0)
        extract_pattern(Composite,"Composite",Jdraw,0)
        extract_pattern(Decorator,"Decorator",Jdraw,0)
        extract_pattern(Facade,"Facade",Jdraw,3)
        extract_pattern(Flyweight,"Flyweight",Jdraw,3)
        extract_pattern(Singleton,"Singleton",Jdraw,3)
        extract_pattern(Proxy,"Proxy",Jdraw,0)
        #extract_pattern(FactoryM,"Factory_Method",Jdraw,0)


        DBPattenV05=pd.DataFrame()
        DBPattenV05['Metrics']=DBPatternX
        DBPattenV05['Patterns']=DBPatternY

        DBPattenV05.sample(frac=1)

        DBPattenV05.to_pickle(pkl)

        print(DBPattenV05["Patterns"].value_counts())
```
